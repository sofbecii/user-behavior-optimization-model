```python
import pandas as pd
import numpy as np

# Tu código previo
general = "E:/Programas/VSCode/resultados_encuesta.xlsx"
hombres = "E:/Programas/VSCode/resultados_encuesta_hombres.xlsx"
mujeres = "E:/Programas/VSCode/resultados_encuesta_mujeres.xlsx"
df_grl = pd.read_excel(general)
df_hom = pd.read_excel(hombres)
df_muj = pd.read_excel(mujeres)


def calcular_estadisticas_columnas(dataframe, lista_columnas):
    """
    Calcula e imprime la media y desviación estándar de las columnas especificadas.
    """
    print(f"{'Columna':<60} | {'Media':<10} | {'Desv. Estándar':<10}")
    print("-" * 90)
    
    for col in lista_columnas:
        if col in dataframe.columns:
            media = dataframe[col].mean()
            std = dataframe[col].std()
            print(f"{col:<60} | {media:<10.2f} | {std:<10.2f}")
        else:
            print(f"⚠️ La columna '{col}' no se encontró en el archivo.")

# Lista de las columnas que proporcionaste
columnas_interes = [
    "Desafío_Importancia", "Desafío_Calificación",
    "Retroalimentación_Importancia", "Retroalimentación_Calificación",
    "Inmersión_Importancia", "Inmersión_Calificación",
    "Concentración_Importancia", "Concentración_Calificación",
    "Claridad de objetivos_Importancia", "Claridad de objetivos_Calificación",
    "Autonomía_Importancia", "Autonomía_Calificación",
    "Interacción social_Importancia", "Interacción social_Calificación",
    "Mejora del conocimiento_Importancia", "Mejora del conocimiento_Calificación",
    "Involucramiento emocional_Importancia", "Involucramiento emocional_Calificación",
    "Equilibrio entre habilidades y tareas_Importancia", "Equilibrio entre habilidades y tareas_Calificación",
    "Narrativa atractiva_Importancia", "Narrativa atractiva_Calificación",
    "Estructura de progresión_Importancia", "Estructura de progresión_Calificación",
    "Animación y sonido_Importancia", "Animación y sonido_Calificación",
    "Calificación Global"
]

# Ejecutar la función
print('\n', 'Análisis general:', '\n')
calcular_estadisticas_columnas(df_grl, columnas_interes)
print('\n', 'Análisis en hombres:', '\n')
calcular_estadisticas_columnas(df_hom, columnas_interes)
print('\n', 'Análisis en mujeres:', '\n')
calcular_estadisticas_columnas(df_muj, columnas_interes)
```
